# Frivolous Toy

## Introduction

I recently decided to explore the AI world of Google Antigravity, and whipped up a toy Javascript app that takes your gender and birth date, and tells you the expected seconds and days of remaining (expected) life.  It uses the latest Australia life tables.

Try it [here](extra/mortality.html)
