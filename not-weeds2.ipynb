# Welcome to the Not-weeds Page 2
## Introduction
I participate in dune and bushcare restoration activities, covering areas from Stumers Creek north of Coolum Beach south to Twin Waters/North Shore on the Maroochy River.


I have a blog page that shows records of weeds seen, being scans of a weeding log-book I (briefly) maintained for a while.

Then I realized that while I was concentrating on learning the scientific species names of the weeds I saw, I had been ignoring the Australian native plants.  So I set myself the goal of learning one Latin scientific name a week, documentating with photos taken during my weeding stints.  Caution:  assigment on names of the species is my own, guided by the Council specialists, and iNaturalist.

The previous page covering some of the native plants is https://coolum001.github.io/notweeds.html

-----------

## Corkwood (Duboisia myoporoides)

The photos below were taken of a single young tree on the sea-facing high dune on my patch.  I was originally surprised to identify it, as some books say it is a rainforest margins species.  However I later found a more mature speciman growing a bit further south (but still in Coolum), so I guess it gets enough rainfall to be happy.

## Horsetail She-oak (Casuarina equisetifolia)

Thriving on the fore-dune, I think that at least some of these she-oaks were originally planted as part of a regeneration effort. Quite a few small she-oaks are coming up, so the effort was successful.

### Images

#### Barrier planting
The she-oaks filter out the salt spray to protect inland species. Licking the leaves give a very salty taste!

## Pigweed (Carpobrotus glaucescens)

Where I gew up, __pigweed__ was the name associated with this species.

The name _Carpobrotus_ indicates that this plant is edible, but I have never been brave enough to try.

## Macaranga tanarius

This is a very vigorous pioneer species

I have a number of these growing in my home garden: very easy to grow from seed, very tolerant of pruning, quick growing shade or privacy screen trees
