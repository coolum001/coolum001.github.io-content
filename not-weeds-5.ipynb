# Welcome to another Not-weeds Page
## Introduction
I participate in dune and bushcare restoration activities, covering areas from Stumers Creek north of Coolum Beach south to Twin Waters/North Shore on the Maroochy River.


I have a blog page that shows records of weeds seen, being scans of a weeding log-book I (briefly) maintained for a while.

Then I realized that while I was concentrating on learning the scientific species names of the weeds I saw, I had been ignoring the Australian native plants.  So I set myself the goal of learning one Latin scientific name a week, documentating with photos taken during my weeding stints.  Caution:  assigment on names of the species is my own, guided by the Council specialists, and iNaturalist.

-----------

## Wollastonia Biflora
 

### Images

I previous post, I promised a photo of the flowers (the Biflora part of the name).  Indeed the flowers come in pairs.

<img src="images/woll-01.jpg" alt="Wallastonia" max-width="90%" max-height="909vh" height="auto"  border= "2px solid #333" padding="15px">


This is a view of sea-facing side of the dune (under open canopy) with Wollastonia covering Asparagus Fern (Asparagus aethiopicus).  In this view, it looks like Wollastonia is winning, but I am deeply pessimistic, that come a long hot dry spell, the water saving adaptions of the Asparagus aethiopicus will win out, and the Wollastonia will shrivel up - all the more reason to get the Asparagus aethiopicus out now!

## Varanus varius (Lace Monitor)
