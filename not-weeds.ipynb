# Welcome to the Not-weeds Page
## Introduction
I participate in dune and bushcare restoration activities, covering areas from Stumers Creek north of Coolum Beach south to Twin Waters/North Shore on the Maroochy River.


I have a blog page that shows records of weeds seen, being scans of a weeding log-book I (briefly) maintained for a while.

Then I realized that while I was concentrating on learning the scientific species names of the weeds I saw, I had been ignoring the Australian native plants.  So I set myself the goal of learning one Latin scientific name a week, documentating with photos taken during my weeding stints.  Caution:  assigment on names of the species is my own, guided by the Council specialists, and iNaturalist.

-----------

## Dune Fan Flower (Scaevola calendulacea)

The photos below were taken in the National Park to the north of Stumers Creek mouth (where we weed once a month).  I don't remember seeing Scaaevola calendulacea in our usual patch, that maybe because there are no weeds where it grows, so we don't go there, so I don't notice it  (this is not by any means a thorough species diversity survey!).  However, I think our usual patch as a slightly different canopy structure with shade trees growing quite close to the frontal dune, whereas these plants were in full sun, with no canopy providing shade.

There was a program of planting _casuarina equisetifolia_ along the frontal coastal dune closer to Coolum, and that may have deterred full-sun ground covers?.

## Wollastonia Biflora

One of the heartening things about weeding is ripping out a thick carpet of Asparagous Fern, and coming back later to see Wollastonai and Dianella covering the area.  Certainly we see more wallabies when Wollastonia replaces Asparagous Fern.  It seems to do well in partial shade, but in roadworks that repaired a landslide just to the south of Coolum, Wollstania is going gang-busters in the full sun.

## Banksia Robur (Swamp Banksia)

The images below are of a Banksia seen on pathway behind coastal dune: low spreading shrub, maybe regrowth from stump, mixed in with other plants. Leaves serrated, flower dark brown.

_Banksia Robur_ always seem a little ratty to me: leaves mottled, flowers turn to a dingy brown, shrub format untidy.  Still, one feature of the other, larger mature Banksia trees is that they seem very prone being blown over in the strong on-shore winds. Given that a major fire went through this area some time in the 1990-s, maybe 30 years is the life-expectancy of larger Banksia trees (usually _Banksia integrifolia_).  I don't think I'll live long enough to see a stable climax (if there is one for coastal dune communities: apparently nutriment leaching eventually causes environment decline, even if the area was untouched by fires, coastal erosian, climate change, or introduced species)

## Pandanus tectorius

The iconic Pandanus palm: grows well on the sea-side slope of the first major dune, and occasionally deeper into the hind-dune forest.  

I was interested to learn that the Pandanus palm was key species for Polynesian sail making: some attribute the lack of Pandanus palms (cooler in NZ) for sail making as part of the explanation for the Maori not expanding from New Zealand.
