# Welcome to the Not-weeds (and some weeds) Page 4
## Introduction
I participate in dune and bushcare restoration activities, covering areas from Stumers Creek north of Coolum Beach south to Twin Waters/North Shore on the Maroochy River.


I have a blog page that shows records of weeds seen, being scans of a weeding log-book I (briefly) maintained for a while.

Then I realized that while I was concentrating on learning the scientific species names of the weeds I saw, I had been ignoring the Australian native plants.  So I set myself the goal of learning one Latin scientific name a week, documentating with photos taken during my weeding stints.  Caution:  assigment on names of the species is my own, guided by the Council specialists, and iNaturalist.

A previous page covering some of the native plants is https://coolum001.github.io/notweeds.html

Having recently attend a lecture and field walk run by Rusty Linnane, the author of the excellant book __Weedlings and Seedlings: A Field Guide to Seedling Identification for Bush Regeneration__, expect to see more photos of small seedlings

## Flame tree (Brachychiton acerifolius)

Small unusual sapling growing at top of a coastal dune under banksia and casuarina canopy (iNaturalist says it is  Brachychiton acerifolius)

This may represent a trend: I have seen another one of these (much larger) in the coastal area, but it is really a North Queensland species. 
There is some conjecture that climate change (increased humidity, etc) may be making it more friendly for tropical species down here.  
Almost certainly, the seeds for this came from a suburbian garden

## Tape vine (Stephania japonica )

Small seedling growing on landward slope of coastal dune under open canopy. iNaturalist agrees it is Stephania japonica 

Notice each leave grows from a stalk, and is not directly attached to the vine.  This distinguishes it from Passiflora suberosa (Corky Passion), which __is__ a weed
