# Ulam sequences and Typst

## Introduction

I recently ran across two areas of interest:

- Ulam numbers
- The [Typst](https://typst.app/home) (sort-of) replacement for Tex, for producing publication ready, math heavy, documents.

Seeing how Ulam sequences present a challenge for efficient computation, I though I would use Typst to generate a PDF describing my attempts to rise to the challenge.  The PDF is [here](extra/ulam.pdf)

## Conclusions

Typst is certainly relatively easy to use, and I could achieve what I wanted in the area of displaying mathematical related diagrams.  It even created nice graphs for me.

I will certainly be trying it out for future documents.  Sadly, the HTML generation is still a work in progress, and not quite there for what I want.
